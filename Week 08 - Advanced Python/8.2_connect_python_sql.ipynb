```python
import pandas as pd                              # For data manipulation and displaying query results
import pymysql                                   # MySQL driver used by SQLAlchemy
from sqlalchemy import create_engine, text       # Core SQLAlchemy functions
import getpass                                   # To securely get password without showing it
```

### SQLAlchemy

#### - What it is:
    - SQLAlchemy is a Python toolkit and ORM (Object Relational Mapper) for interacting with SQL databases
    - It allows Python programs to talk to databases like MySQL, PostgreSQL, SQLite, etc., without writing unsafe raw SQL all the time

### PyMySQL

#### What it is:

    - PyMySQL is a pure-Python MySQL client library
    - It allows Python programs to communicate with a MySQL database 

#### Why we need it:

    - SQLAlchemy itself doesn’t talk to MySQL directly
    - SQLAlchemy uses drivers like PyMySQL, MySQLdb, or mysqlclient

```python
db = "sakila"

connection_string = 'mysql+pymysql://root:' + password + '@localhost/' + db
engine = create_engine(connection_string)
engine
```

```python
with engine.connect() as connection:
    query = text('SELECT * FROM actor')
    result = connection.execute(query)
    df = pd.DataFrame(result.all())
```

```python
df
```

##### - text allows us to write raw SQL queries safely

#### - engine.connect() opens a connection to the database.

    - The engine itself is just a Python object that knows how to talk to the database
    - It does not actually open a connection until you call connect()
    - connect() gives you a connection object that you can use to execute queries, fetch data, or commit changes

#### - with engine.connect() as connection:

    - Opens a connection to the database
    - Automatically closes it at the end of the block

#### - query = text('SELECT * FROM actor')

    - Defines a raw SQL query
    - text() ensures SQLAlchemy treats it correctly and allows parameter binding if needed

#### - result = connection.execute(query)
    
    - Executes the query against the database
    - Returns a ResultProxy object that contains all rows

#### - df = pd.DataFrame(result.all(), columns=result.keys())

    - Converts the result to a Pandas DataFrame for easy viewing
    - result.all() fetches all rows 
    - result.keys() gives the column names

- Summary of Steps

    - 1- Import necessary libraries (pandas, pymysql, sqlalchemy, getpass)
    - 2- Get the password securely
    - 3- Specify the database
    - 4- Create the connection string and engine
    - 5- Use with engine.connect() to open a connection 
    - 6- Use text() to define queries 
    - 7- Execute queries with connection.execute() 
    - 8- Convert results to a Pandas DataFrame for easy viewing 
    - 9- The connection closes automatically at the end of the with block 

```python
def rentals_month(engine, month, year):
    with engine.connect() as connection:
        txt = f'''SELECT *, YEAR(rental_date) AS YEAR, MONTH(rental_date) AS MONTH
                  FROM sakila.rental
                  WHERE YEAR(rental_date) = {year} AND MONTH(rental_date) = {month};'''
        query = text(txt)
        result = connection.execute(query)
        df_1 = pd.DataFrame(result.all())
        return df_1
```

```python
rentals_month(engine, 5 , 2005)
```

```python
db = "shop_project"

connection_string = 'mysql+pymysql://root:' + password + '@localhost/' + db
engine = create_engine(connection_string)
engine
```

```python
def sign_up(engine, customer_name, customer_email, customer_pass):
    with engine.connect() as connection:
        txt = f'''INSERT INTO shop_project.customers (Customer_Name, Email, Password)
                  VALUES ("{customer_name}", "{customer_email}", "{customer_pass}");'''
        query = text(txt)
        connection.execute(query)
        connection.commit() # NEEDED WHEN UPDATING THE TABLE, commits changes to databse
        return print("sign up successful")
```
