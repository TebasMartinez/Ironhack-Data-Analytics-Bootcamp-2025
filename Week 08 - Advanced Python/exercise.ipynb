### Same without pandas 

```python
sold_prod_ids = []
sold_prod_qtys = []
for order in coffee_shop["orders"]:
    for item in (order["items"]):
        sold_prod_ids.append(item["product_id"])
        sold_prod_qtys.append(item["qty"])
```

```python
prod_ids = []
prod_prices = []
prod_names = []
for product in coffee_shop["products"]:
    prod_ids.append(product["id"])
    prod_prices.append(product["price"])
    prod_names.append(product["name"])
```

```python
sold_prod_total_prices = {}
for ind2, prod in enumerate(prod_ids):
    for ind1, sold_prod in enumerate(sold_prod_ids):
        if sold_prod == prod:
            total = sold_prod_qtys[ind1] * prod_prices[ind2]
            if prod not in sold_prod_total_prices:
                sold_prod_total_prices[prod] = total
            else:
                sold_prod_total_prices[prod] += total
```

```python
most_sold_prod_id = max(sold_prod_total_prices, key=sold_prod_total_prices.get) 
```

```python
for ind, prod in enumerate(prod_ids):
    if prod == most_sold_prod_id:
        print(prod_names[ind])
```

## Total revenue of our shop in Berlin

```python
exploded_orders = orders.explode("items")
```

```python
berlin_customer_ids = customers[customers["city"] == "Berlin"]["id"]
```

```python
berlin_orders = exploded_orders[exploded_orders["customer_id"].isin(berlin_customer_ids)]
```

```python
berlin_orders
```

```python
berlin_total = 0
for product in berlin_orders["items"]:
    product_price = float(products[products["id"] == product["product_id"]]["price"])
    total_product = product_price * product["qty"]
    berlin_total += total_product
```
