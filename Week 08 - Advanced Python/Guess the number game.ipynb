Create a game where the user provides a maximum number and a number of attempts possible. Computer generates random number between 1 and maximum number. User has to try to guess the number. Computer says if number is too high, too low, or right.

```python
import random
```

```python
def game():
    # 1. Let user pick the range
    print("Welcome to Guess the number!")
    try:
        number_range = int(input("Please choose the range you'd like to play with. Computer can choose a number from 1 to: "))
    except ValueError:
        print("Please enter an integer number as the top of the range!")
    
    # 2. Computer picks a random number
    number = random.choice(range(1, number_range+1))
    
    # 3. Ask user to choose max attempts
    try:
        attempts = int(input("Plese choose the number of attempts you'd like to have: "))
    except ValueError:
        print("Please enter an integer number as the number of attempts!")
        
    # 4. Star game:
    guess = 0
    print("Computer has chosen a number")
    while guess != number and attempts != 0:
        # User guesses, and either we say "guess is too low" or "guess is too high"
        try:
            guess = int(input("Please enter your guess: "))
            if not isinstance(guess, int):
                raise ValueError("Please enter an integer as your guess!")
        except ValueError as ve:
            print(ve)
        if guess != number:
            if guess > number:
                print("That's not the number Computer chose. Your number is TOO HIGH. Please try again!")
            else:
                print("That's not the number Computer chose. Your number is TOO LOW. Please try again!")
            attempts -= 1
            print(f"You have {attempts} attempts left.")
    # If the guess is correct, user wins
    if guess == number:
        print(f"You win! Computer chose the number {number} and you guessed it! :D")
    else:
        print(f"You lose :( Computer had chosen the number {number}. Do better next time...")
    
    # Make sure you keep track of the attempts
```

```python
game()
```
