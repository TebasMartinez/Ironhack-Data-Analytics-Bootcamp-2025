```python
url = "https://en.wikipedia.org/wiki/List_of_presidents_of_the_United_States"
headers = {
       "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
   }
response = requests.get(url, headers=headers)
print(response)
soup = BeautifulSoup(response.content)
base_url = "https://en.wikipedia.org"

presidents = []
vicepresidents = []
born_info = []
for link in soup.select("td b a"):
    new_url = base_url + link["href"]
    response = requests.get(new_url, headers=headers)
    soup = BeautifulSoup(response.content)
    presidents.append(soup.select("tr th.infobox-above div")[0].get_text())
    vicepresidents.append(soup.find_all("span", string = "Vice President")[0].parent.parent.select("td")[0].get_text().strip().replace("\n", " / "))
    born_info.append(soup.find_all("th", string = "Born")[0].parent.select("td")[0].get_text(" ", strip=True).encode('ascii', 'replace').decode("utf-8").replace("?", " "))

president_info = pd.DataFrame({
    "president":presidents,
    "vicepresident":vicepresidents,
    "born":born_info
})
president_info
```

```python
for info in born_info:
    print(info)
```
