```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
def scrape_imdb(title_type, user_rating, start_date, end_date):
    # Build URL
    base_url = "https://www.imdb.com/search/title/"
    url = base_url + "?title_type=" + title_type
    url = url + "&release_date=" + start_date + "," + end_date
    url = url + "&user_rating=" + user_rating

    # Load content
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"}
    response = requests.get(url, headers=headers)
    print(response)
    soup = BeautifulSoup(response.content)

    movie_n = []
    movie_titles = []
    for title in soup.select("a h3.ipc-title__text "):
        title_list = title.get_text().split()
        number = title_list[0]
        mtitle = " ".join(title_list[1:])
        movie_n.append(number)
        movie_titles.append(mtitle)

    years = []
    for year in soup.select("span.sc-15ac7568-7:nth-child(3n + 1)"):
        years.append(year.get_text())

    ratings = []
    for rating in soup.select("span.ipc-rating-star--rating"):
        ratings.append(rating.get_text())

    runtimes = []
    for runtime in soup.select("span.dli-title-metadata-item:nth-child(2n)"):
        time_list = runtime.get_text().split()
        if len(time_list) == 2:
            total = (int(time_list[0].split("h")[0]) * 60) + int(time_list[1].split("m")[0])
        else:
            total = int(time_list[0].split("h")[0]) * 60
        runtimes.append(total)

    descriptions = []
    for desc in soup.select("div.ipc-html-content-inner-div"):
        descriptions.append(desc.get_text())

    votes = []
    for vote_n in soup.select("span.ipc-rating-star--voteCount"):
        votes.append(vote_n.get_text().replace("(","").replace(")",""))

    imdb_search_result = pd.DataFrame ({
        "Movie Nr":movie_n,
        "Title":movie_titles,
        "Year":years,
        "Rating":ratings,
        "Runtime (min)":runtimes,
        "Description":descriptions,
        "Votes":votes
    })
    
    return imdb_search_result
```

```python
df = scrape_imdb("feature", "7.5", "1990", "1992")
df
```
