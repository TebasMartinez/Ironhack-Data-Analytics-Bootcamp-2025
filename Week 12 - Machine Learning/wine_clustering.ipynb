```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_wine
```

## Import data

```python
X, y = load_wine(as_frame=True, return_X_y=True)
```

```python
X
```

```python
y
```

## Scale data

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
scaler.fit(X)
X_scaled = scaler.transform(X)
```

```python
X_scaled
```

## Create model

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
```

```python
kmeans.fit(X_scaled)
```

## Predict clusters

```python
kmeans.predict(X_scaled)
```

```python
X['cluster'] = kmeans.predict(X_scaled)
```

```python
X
```

## Predict a new wine and suggest a similar wine (one in the same cluster)

```python
new_wine = []
for i in X.columns:
    new_wine.append(float(input(f"Enter a value for {i}: ")))
```

```python
new_wine
```

```python
new_wine_array = np.array(new_wine)
```

```python
new_wine_array = new_wine_array.reshape(1,-1)
```

```python
new_wine_array.shape
```

```python
new_wine_scaled = scaler.transform(new_wine_array)
```

```python
kmeans.predict(new_wine_scaled)
```
